#### repartetion & coalesce

```python
youtube_df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("/Volumes/quant_databricks/batch0506/quantcloudrawdatasets/youtube.csv")
)
youtube_df.display()
```

```python
# 128 MB
# 5000 rows -> 1024 MB -> 1024/128 = 8
```

```python
# 5000 -> 1 MB -> 4 partetitions
```

```python
from pyspark.sql import functions as F
youtube_df.select(F.spark_partition_id()).distinct().count()
```

```python
# Repartition
# used for either incresing or decresing the no.of partitions in a df
# in order to do this repartition will perform a full shuffel of the data
```

```python
youtube_df = youtube_df.repartition(4)
youtube_df.select(F.spark_partition_id()).distinct().count()
```

```python
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
youtube_df = youtube_df.repartition(2)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
youtube_df = youtube_df.repartition(5)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
# coalesce
# This function is used only to reduce/decrese the no.of partitions in a dataframe

# coalesce is faster then repartetion because it does not involve in reshuffling of the data as it will try to combine the data from the nearest partitions
```

```python
youtube_df = youtube_df.coalesce(2)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.databricks.adaptive.autoOptimizeShuffle.enabled", "false")
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "false")
```

```python
youtube_df = youtube_df.coalesce(3)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
youtube_df.select(F.spark_partition_id()).distinct().count()
```

```python
youtube_df.coalesce(5).groupBy(F.spark_partition_id()).count()
```

```python
youtube_df = youtube_df.coalesce(10)
youtube_df.select(F.spark_partition_id()).distinct().count()
```

```python
youtube_df = youtube_df.repartition(10)
youtube_df.select(F.spark_partition_id()).distinct().count()
```

```python
youtube_df = youtube_df.repartition(20)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
youtube_df = youtube_df.coalesce(1)
youtube_df.groupBy(F.spark_partition_id()).count().display()
```

```python
# 2 partitions into 10 partitions 
youtube_df = youtube_df.coalesce(10)  # no impact
youtube_df.groupBy(F.spark_partition_id()).count().display()
```
