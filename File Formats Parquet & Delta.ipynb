```python
import re

invalid_chars = r"[ ,;{}()\n\t=%]"
new_columns = [re.sub(invalid_chars, "", col).lower() for col in youtube_df.columns]
youtube_df = youtube_df.select(*[F.col(old_col).alias(new_col) for new_col, old_col in zip(new_columns, youtube_df.columns)])
display(youtube_df)
```

```python
target_base_path = "/Volumes/quant_databricks/quant-learning/uc-quant-learning"
youtube_df.write.format("delta").mode("overwrite").save(f"{target_base_path}/youtube_delta")
```
