```python
# Method 1 - With out strict schema 

emp_data = [
    (100, "Naveen", 20000, "IT"),
    (200, "Ravi", 30000, "HR"),
    (300, "Rajesh", 40000, "IT"),
    (400, "Ramesh", 50000, "IT"),
    (500, "Rakesh", 60000, "IT"),
    (600, "Raj", 70000, "IT"),
    (700, "Rajesh", 80000, "IT"),
    (800, "Ravi", 90000, "IT"),
    (900, "Rajesh", 100000, "HR")
]

cols = ["id", "name", "salary", "dept"]

emp_df = spark.createDataFrame(data=emp_data, schema=cols)
display(emp_df)
```

```python
from pyspark.sql import functions as F
from pyspark.sql import types as T
```

```python
# Method 2 - with strict schema enforced

emp_data = [
    (100, "Naveen", 20000, "IT"),
    (200, "Ravi", 30000, "HR"),
    (300, "Rajesh", 40000, "IT"),
    (400, "Ramesh", 50000, "HR"),
    (500, "Rakesh", 60000, "IT"),
    (600, "Raj", 70000, "HR"),
    (700, "Rajesh", 80000, "IT"),
    (800, "Ravi", 90000, "IT"),
    (900, "Rajesh", 100000, "HR"),
    (1000, None, 35000, "HR")
]

emp_schema = T.StructType([
    T.StructField("id", T.IntegerType(), False),
    T.StructField("name", T.StringType(), True),
    T.StructField("salary", T.IntegerType(), True),
    T.StructField("dept", T.StringType(), True)
])


emp_df = spark.createDataFrame(data=emp_data, schema=emp_schema)
display(emp_df)
```

```python
# Selecting columns from the dataframe

display(emp_df.select("id", "name"))
```

```python
display(emp_df.select(emp_df.id, emp_df.name))
```

```python
display(emp_df.select(emp_df["id"], emp_df["name"]))
```

```python
display(emp_df.select(F.col("id"), F.col("name")))
```

```python
display(emp_df.select("*"))
```

```python
# Filter the data from df
filtered_empdf = emp_df.filter(F.col("dept") == "HR")
display(filtered_empdf)
```

```python
filtered_empdf = emp_df.where(F.col("dept") == "HR")
display(filtered_empdf)
```

```python
# sql like select expression

display(emp_df.selectExpr("id", "salary as emp_salary"))
```

```python

display(emp_df.selectExpr("id", "salary as emp_salary", "salary * 0.10 as bonus"))
```

```python

display(emp_df.selectExpr("id", "upper(name) as emp_name", "salary as emp_salary", "salary * 0.10 as bonus"))
```
