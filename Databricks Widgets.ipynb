```python
dbutils.widgets.text("My Widget Name", "QCT")
```

```python
value = dbutils.widgets.get("My Widget Name")
print(value)
```

```python
dbutils.widgets.dropdown("MyDropDownWidget", "Option1", ["Option1", "Option2", "Option3"])
dropvalue = dbutils.widgets.get("MyDropDownWidget")
print(dropvalue)
```

```python
dbutils.widgets.multiselect("MyMultiWidget", "Option1", ["Option1", "Option2", "Option3"])
multivalue = dbutils.widgets.get("MyMultiWidget").split(",")
print(multivalue)
```

```python
dbutils.widgets.combobox("MyComboBoxWidget", "Option1", ["Option1", "Option2", "Option3"])
comboboxvalue = dbutils.widgets.get("MyComboBoxWidget")
print(comboboxvalue)
```
