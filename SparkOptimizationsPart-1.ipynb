#### Spark Broadcast Join

```python
spark.conf.set("spark.sql.adaptive.enabled", "false")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "-1")
```

```python
from pyspark.sql import functions as F
```

```python
# Students

stu_data = [
    (1, 'John', 'Doe', 'john.doe@example.com', 200),
    (2, 'Jane', 'Doe', 'jane.doe@example.com', 100),
    (3, 'Alice', 'Smith', 'alice.smith@example.com', 500),
    (4, 'Bob', 'Johnson', 'bob.johnson@example.com', 400),
    (5, 'Charlie', 'Williams', 'charlie.williams@example.com', 200),
    (6, 'David', 'Brown', 'david.brown@example.com', 100),
    (7, 'Eve', 'Davis', 'eve.davis@example.com', 300),
    (8, 'Frank', 'Miller', 'frank.miller@example.com', 400),
    (9, 'Grace', 'Taylor', 'grace.taylor@example.com', 500),
    (10, 'Hannah', 'Anderson', 'hannah.anderson@example.com', 300)
]

columns = ['stu_id', 'first_name', 'last_name', 'email', 'dep_id']
# Create a DataFrame from the data and columns
stu_df = spark.createDataFrame(stu_data, columns)
display(stu_df)
```

```python
dep_data = [
    (100, 'Computer Science', 'Engineering'),
    (200, 'Mathematics', 'Science'),
    (300, 'Physics', 'Science'),
    (400, 'Chemistry', 'Science'),
    (500, 'Biology', 'Science')
]

dep_cols = ['dep_id', 'name', 'category']
dep_df = spark.createDataFrame(dep_data, dep_cols)
display(dep_df)
```

```python
# perform a simple join between students and dep tables

result_df = (
    stu_df
    .join(
        dep_df,
        on=["dep_id"],
        how="inner"
    )
)

result_df.display()
```

```python
# Spark Joins

# default join stratagy is sort merge join
# broadcast join -- optimized join
# shuffel hash join
```

```python
# Broadcast Join

result_df = (
    stu_df
    .join(
        F.broadcast(dep_df),
        on=["dep_id"],
        how="inner"
    )
)

result_df.display()
```

```python
spark.conf.set("spark.sql.adaptive.enabled", "true")
spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "10MB")
```
